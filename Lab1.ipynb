#### <div class="alert alert-block alert-success"> 1.4 Vectors </div>
A vector is one of the most basic data structures in R.

A vector stores multiple values of the same general type.

````r
# Creating a vector
vec <- c(1, 2, 3, 4, 5)
vec
````

#### <div class="alert alert-block alert-success"> 1.6 Matrices </div>

Matrices are important in data science because many machine learning methods are built on matrix operations.

````r
A <- matrix(
  c(1, 2, 3, 4, 5, 6),
  nrow = 2,
  ncol = 3
)

A
````

````r
dim(A)
````

#### <div class="alert alert-block alert-success"> 1.7 Data Frames </div>

A data frame is one of the most important data structures in R.

A data frame is like a spreadsheet:

- Each row is an observation.
- Each column is a variable.

For example, suppose we have data from several cars.

````r
cars_small <- data.frame(
  car = c("Car A", "Car B", "Car C", "Car D"),
  weight = c(2200, 2800, 3200, 3600),
  mpg = c(32, 27, 23, 19)
)

cars_small
````

Here:

- Each row represents one car.
- `weight` is the vehicle weight in pounds.
- `mpg` is fuel efficiency, measured in miles per gallon.

````r
str(cars_small)
````

````r
summary(cars_small)
````

````r
cars_small$weight
````

````r
cars_small$mpg
````

The `$` symbol selects one column from a data frame.

#### <div class="alert alert-block alert-success"> 1.8 Tibbles </div>

Tibbles print nicely and are convenient for data analysis.

````r
cars_tibble <- as_tibble(cars_small)

cars_tibble
````

#### <div class="alert alert-block alert-success"> 1.9 Data Tables </div>
A `data.table` is another data structure for working with data in R.

It is especially useful for large datasets because it is fast and memory-efficient.

````r
# Install if needed:
# install.packages("data.table")

library(data.table)
````

````r
cars_dt <- as.data.table(cars_small)

cars_dt
````

Basic `data.table` syntax has the form:

```r
DT[i, j, by]
```

where:
- `i` selects rows
- `j` selects columns or calculations
- `by` groups data

````r
# Select cars with weight greater than 2800
cars_dt[weight > 2800]
````

````r
# Compute average MPG
cars_dt[, mean(mpg)]
````

````r
# Select only weight and mpg columns
cars_dt[, .(weight, mpg)]
````

In this camp, we will use both `tidyverse` and `data.table`.

- `tidyverse` is readable and widely used for teaching.
- `data.table` is fast and powerful for larger datasets.

You do not need to master both today. The goal is to become comfortable reading and running basic commands.

#### <div class="alert alert-block alert-success"> 3.1 Checking for Missing Values </div>

Missing values are represented by `NA` in R.

One of the first things a data scientist should do is determine whether any data are missing.

````r
sum(is.na(stops))
````

````r
colSums(is.na(stops))
````

Questions

1. Which variables contain missing values?
2. Why might information be missing?
3. Should we remove those observations or keep them?

````r
missing_summary <- data.table(
  variable = names(stops),
  missing = colSums(is.na(stops))
)

missing_summary[, percent_missing :=round(100 * missing / nrow(stops), 2)]

missing_summary[order(-percent_missing)]
````

Should we trust a variable with 50% missing values?

#### <div class="alert alert-block alert-success"> 3.2 Convert Variables </div>

````r
stops[ , subject_sex := factor(subject_sex)]

stops[ , subject_race := factor(subject_race)]

stops[ , violation := factor(violation)]
````

#### <div class="alert alert-block alert-success"> 4.1 The Grammar of Graphics </div>

Most ggplot figures follow the same structure:

```r
ggplot(data, aes(...)) +
    geom_...
```

Think of this as:

1. Choose a dataset
2. Choose variables
3. Choose a type of plot

#### <div class="alert alert-block alert-success"> 4.2 Distribution of Driver Age </div>

A histogram shows how numerical values are distributed.

Question:

How old are the drivers in our dataset?

````r
ggplot(stops_clean, aes(subject_age)) +
  geom_histogram(bins = 20) +
  labs(
    title = "Distribution of Driver Age",
    x = "Age",
    y = "Count"
  ) +
  theme_minimal()
````

Discussion

- What age range appears most common?
- Are very young drivers common?
- Are very old drivers common?

#### <div class="alert alert-block alert-success"> 4.3 Most Common Violations </div>

Bar charts are useful for categorical variables.

Question:

Which violations occur most often?

````r
top_violation <- stops_clean[ , .N, by = violation][order(-N)][1:10]

ggplot(top_violation,  aes(x = reorder(violation, N), y = N)) +
  geom_col() +
  coord_flip() +
  labs(
    title = "Top 10 Most Common Violations",
    x = "Violation",
    y = "Number of Stops"
  ) +
  theme_minimal()
````

````r
top_violation[ , violation_short := substr(violation, 1, 10)]

ggplot(top_violation, aes(x = reorder(violation_short, N), y = N)) +
  geom_col() +
  coord_flip() +
  labs(
    title = "Top 10 Most Common Violations",
    x = "Violation",
    y = "Number of Stops"
  ) +
  theme_minimal()
````

What proportion of all stops come from the top 10 violations?

````r
top10_count <- top_violation[ , sum(N)]

total_count <- nrow(stops_clean)

100 * top10_count / total_count
````
